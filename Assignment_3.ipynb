# **In lab Practise Task 1**

```python
import random

class Environment:
    def __init__(self):
        self.state = random.choice(['Clean', 'Dirty'])

    def get_percept(self):
        return self.state

    def clean_room(self):
        self.state = 'Clean'

    def change_state(self):
        self.state = random.choice(['Clean', 'Dirty'])


class SimpleReflexAgent:
    def __init__(self):
        pass

    def act(self, percept):
        if percept == 'Dirty':
            return 'Clean the room'
        else:
            return 'Room is clean'


def run_agent(agent, environment, steps):
    for step in range(steps):
        percept = environment.get_percept()
        action = agent.act(percept)
        print(f"Step {step + 1}: Percept - {percept}, Action - {action}")
        if percept == 'Dirty':
            environment.clean_room()
        environment.change_state()


# Create instances of agent and environment
agent = SimpleReflexAgent()
environment = Environment()

# Run the agent in the environment for 5 steps
run_agent(agent, environment, 5)
```

# **Task 2** – Smart Classroom Light Controller (Model-Based Agent)

```python
import random

class Environment:
    def __init__(self):
        self.students_present = random.choice(['Yes', 'No'])
        self.light_status = random.choice(['ON', 'OFF'])

    def get_percept(self):
        return self.students_present

    def get_light_status(self):
        return self.light_status

    def action(self, act):
        if act == 'ON':
            self.light_status = 'ON'
        elif act == 'OFF':
            self.light_status = 'OFF'

    def update_state(self):
        self.students_present = random.choice(['Yes', 'No'])


class Model_based_light_controller:
    def __init__(self, light):
        self.model={}
        self.model['light_status']=light

    def update_model(self, percept):
      self.model['students_present'] =percept

    def predict_light_status(self):
      if self.model['students_present']=='Yes' and self.model['light_status']=='OFF':
        return 'ON'
      elif self.model['students_present']=='No' and self.model['light_status']=='ON':
        return 'OFF'
      else:
        return ''

    def act(self, percept):
      self.update_model(percept)
      temp=self.predict_light_status()
      if temp!='':
        self.model['light_status']=temp
      return temp

def run_agent(agent, environment, steps):
    for step in range(steps):
        percept = environment.get_percept()
        action = action2 = agent.act(percept)
        if action=='':
          action='No action taken'
        elif action=='ON':
          action='Light turned ON'
        else:
          action='Light turned OFF'
        print(f"Step {step + 1}: Students present - {percept}, Light status - {env.get_light_status()}, Action - {action}")
        environment.action(action2)
        environment.update_state()


env = Environment()
agent = Model_based_light_controller(env.get_light_status())
run_agent(agent, env, 8)
```

# **Task 3** – Student Study Planner (Goal-Based Agent)

```python
class Environment:
    def __init__(self):
        self.unfinished_subjects = []
        self.completed_subjects = []

    def get_percept(self):
        return self.unfinished_subjects

    def study(self, subject):
        if subject in self.unfinished_subjects:
            self.unfinished_subjects.remove(subject)
            self.completed_subjects.append(subject)

    def add_subject(self, subject):
        self.unfinished_subjects.append(subject)

class Goal_based_student_planner:
    def __init__(self):
        self.goal = 'Complete all subjects'

    def formulate_goal(self, percept):
        if (percept==[]):
          return ''
        else:
          return percept[0]

    def act(self, percept):
        x = self.formulate_goal(percept)
        if x == '':
          print('Goal achieved: All subjects completed')
        else:
          print(f'Studying {x}')
        return x

def run_agent(agent, env):
    while True:
        percept = env.get_percept()
        action = agent.act(percept)
        if action == '':
          break
        env.study(action)

agent=Goal_based_student_planner()
env=Environment()
env.add_subject('AI')
env.add_subject('Math')
env.add_subject('Physics')
run_agent(agent, env)
```

# **Task 4** – Restaurant Selector (Utility-Based Agent)

```python

class Environment:
    def __init__(self, restaurants=None, distance=None, rating=None):
        if restaurants is None or distance is None or rating is None or len(restaurants)!=len(distance) or len(restaurants)!=len(rating):
            self.restaurants = ['A', 'B']
            self.distance = [3, 5]
            self.rating = [7, 9]
        else:
          self.restaurants = restaurants
          self.distance = distance
          self.rating = rating

    def get_percept(self):
        return self.restaurants, self.distance, self.rating


class UtilityBasedAgent:
    def __init__(self):
        pass

    def utility(self, distance, rating):
        return rating-distance

    def act(self, percept):
        best_restaurant = None
        best_utility = -float('inf')
        restaurants, distances, ratings = percept
        for i in range(len(restaurants)):
            restaurant = restaurants[i]
            distance = distances[i]
            rating = ratings[i]
            restaurant_utility = self.utility(distance, rating)
            print(f"Restaurant {restaurant} Utility = {restaurant_utility}")
            if restaurant_utility > best_utility:
                best_restaurant = restaurant
                best_utility = restaurant_utility

        return best_restaurant

def run_agent(agent, environment):
    percept = environment.get_percept()
    best_choice = agent.act(percept)
    print(f"Selected Restaurant: {best_choice}")

agent = UtilityBasedAgent()
environment = Environment()

run_agent(agent, environment)
```

# **Task 5** – Learning Agent Game

```python
import random

class LearningBasedAgent:
    def __init__(self):
        self.Q = {'Play': 0.0, 'Rest': 0.0}
        self.actions = ['Play', 'Rest']
        self.alpha = 0.1  # Learning rate
        self.gamma = 0.9  # Discount factor
        self.epsilon = 0.1  # Exploration rate

    def get_Q_value(self, action):
        return self.Q[action]

    def select_action(self):
        if random.uniform(0, 1) < self.epsilon:
            return random.choice(self.actions)
        else:
            return max(self.actions, key=lambda a: self.get_Q_value(a))

    def learn(self, action, reward):
        old_Q = self.get_Q_value(action)
        best_future_Q = max([self.get_Q_value(a) for a in self.actions])
        self.Q[action] = old_Q + self.alpha * (reward + self.gamma * best_future_Q - old_Q)
        print("Q-table updated")

    def act(self):
        action = self.select_action()
        return action

class Environment:
    def __init__(self):
        pass

    def play_reward(self):
        return 5

    def rest_reward(self):
        return 1

def run_agent(agent, environment, steps):
    for step in range(steps):
        action = agent.act()
        if action == 'Play':
            reward = environment.play_reward()
            print(f"Step {step + 1}: Action {action} Reward {reward}")
        else:
            reward = environment.rest_reward()

    agent.learn(action, reward)


# Create instances of agent and environment
agent = LearningBasedAgent()
environment = Environment()

run_agent(agent, environment, 10)
```

# **Task 6** - Firefighting Robot

```python
class Environment:
    def __init__(self):
      self.rooms = [
          'safe', 'safe', 'fire',
          'safe', 'fire', 'safe',
          'safe', 'safe', 'fire'
      ]

    def get_percept(self, position):
      if (position>=0 and position<9):
        return self.rooms[position]

    def action(self, act, position):
      if act == 'extinguish fire':
        self.rooms[position] = 'safe'

    def display(self, position=-1):
      print("\nCurrent Grid State:")
      rooms_with_agent = self.rooms[:]

      for i in range(0, 9):
        if (rooms_with_agent[i]=='safe'):
          rooms_with_agent[i] = "⬜️"
        else:
          rooms_with_agent[i] = "💥"

        if (position>=0 and position <9):
          if (self.rooms[position]=='safe'):
            rooms_with_agent[position] = "🔳"
          else:
            rooms_with_agent[position] = "🟥"
        print(rooms_with_agent[i], end="")
        if (i+1)%3==0:
          print()
      print()

class agent:
  def __init__(self):
    self.location = 0

  def get_position(self):
    return self.location

  def act(self, percept):
    if percept == 'fire':
      return 'extinguish fire'
    else:
      self.location += 1
      if self.location >= 9:
        self.location = 8
        return 'Goal completed'
      return 'go to next room'

def run_agent(agent, environment):
  step=1
  while (agent.get_position()<9):
    position = agent.get_position()
    percept = environment.get_percept(position)
    action = agent.act(percept)
    print(f"Step {step}: Percept - {percept}, Action - {action}")
    environment.display(position)
    environment.action(action, position)
    if (action=='Goal completed'):
      print("END")
      break
    step += 1

  print("\nFinal State")
  environment.display()


agent = agent()
environment = Environment()

run_agent(agent, environment)
```
