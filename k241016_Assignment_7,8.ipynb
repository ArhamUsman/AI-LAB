````python
import math

class Node:
    def __init__(self, value=None):
        self.value=value
        self.children=[]
        self.minmax_val=None

def initialise_minmax():
  global root, n1, n2, n3, n4, n5, n6, n7, n8, n9, n10, n11, n12, n13, n14
  root.minmax_val=None
  n1.minmax_val=None
  n2.minmax_val=None
  n3.minmax_val=None
  n4.minmax_val=None
  n5.minmax_val=None
  n6.minmax_val=None
  n7.minmax_val=None
  n8.minmax_val=None
  n9.minmax_val=None
  n10.minmax_val=None
  n11.minmax_val=None
  n12.minmax_val=None
  n13.minmax_val=None
  n14.minmax_val=None

def standard_minmax(node, depth, max_player=True):
    global count
    count+=1
    if depth==0 or not node.children:
      try:
        val=int(node.value)
        return val
      except:
        print("Error! Leaf node not found")
        if max_player:
          return -math.inf
        return math.inf
    if max_player:
        value= -math.inf
        for child in node.children:
            child_val=standard_minmax(child, depth-1, False)
            value=max(value, child_val)
        node.minmax_val=value
        return value
    else:
        value=math.inf
        for child in node.children:
            child_val=standard_minmax(child, depth-1, True)
            value=min(value, child_val)
        node.minmax_val=value
        return value

root=Node('Root')
n1=Node('N1')
n2=Node('N2')
n3=Node('N3')
n4=Node('N4')
n5=Node('N5')
n6=Node('N6')
n7=Node(4)
n8=Node(7)
n9=Node(2)
n10=Node(5)
n11=Node(1)
n12=Node(8)
n13=Node(3)
n14=Node(6)

root.children=[n1, n2]
n1.children=[n3, n4]
n2.children=[n5, n6]
n3.children=[n7, n8]
n4.children=[n9, n10]
n5.children=[n11, n12]
n6.children=[n13, n14]

print("\nRunning Standard Min Max Algo")
count=0
standard_minmax(root, 3)
print("MinMax Values:")
print(f"root : {root.minmax_val}")
print(f"n1 : {n1.minmax_val}")
print(f"n2 : {n2.minmax_val}")
print(f"n3 : {n3.minmax_val}")
print(f"n4 : {n4.minmax_val}")
print(f"n5 : {n5.minmax_val}")
print(f"n6 : {n6.minmax_val}")

print("\nAdding depth limit 2")
count=0
initialise_minmax()
standard_minmax(root, 2)
print("MinMax Values:")
print(f"root : {root.minmax_val}")
print(f"n1 : {n1.minmax_val}")
print(f"n2 : {n2.minmax_val}")
print(f"n3 : {n3.minmax_val}")
print(f"n4 : {n4.minmax_val}")
print(f"n5 : {n5.minmax_val}")
print(f"n6 : {n6.minmax_val}")
````

# **Task 2: Alpha-Beta Pruning**
1. Implement Minimax with Alpha-Beta pruning on the same tree.

2. Track alpha and beta values.

3. Print:
- - Minimax values for all nodes
- - Nodes that are pruned

4. Compare the number of nodes visited with standard Minimax and explain why
pruning reduces computation.

````python
def alpha_beta_minmax(node, depth, alpha, beta, max_player=True):
    global count
    count+=1
    if depth==0 or not node.children:
        return node.value

    if max_player:
        value= -math.inf
        for child in node.children:
            child_val=alpha_beta_minmax(child, depth-1, alpha, beta, False)
            value=max(value, child_val)
            alpha=max(alpha, value)
            if beta<=alpha:
                break
        node.minmax_val=value
        return value
    else:
        value=math.inf
        for child in node.children:
            child_val=alpha_beta_minmax(child, depth-1, alpha, beta, True)
            value=min(value, child_val)
            beta=min(value, beta)
            if beta<=alpha:
                break
        node.minmax_val=value
        return value

def print_pruned_nodes(node):
    if node.children:
        if node.minmax_val==None:
            print(f"Pruned {node.value}")
            return
        for child in node.children:
            print_pruned_nodes(child)

count=0
initialise_minmax()
standard_minmax(root, 3)
print(f"No of nodes visited by standard minmax: {count}")

initialise_minmax()

print("\nRunning Alpha beta Min Max Algo")
count=0
initialise_minmax()
alpha_beta_minmax(root, 3, -math.inf, math.inf)
print("MinMax Values:")
print(f"root : {root.minmax_val}")
print(f"n1 : {n1.minmax_val}")
print(f"n2 : {n2.minmax_val}")
print(f"n3 : {n3.minmax_val}")
print(f"n4 : {n4.minmax_val}")
print(f"n5 : {n5.minmax_val}")
print(f"n6 : {n6.minmax_val}")

print("\nPrinting nodes that are pruned:")
print_pruned_nodes(root)

print()
print(f"No of nodes visited by alpha beta minmax: {count}")
print()

# In this specific scenario, both visit the same no of nodes
# Pruning reduce computational cost as whole branch generated from that stem is discarded when beta<=alpha so no need to visit the entire branch so less computation Required
````

# **Task 4 Basic CSP Modelling**
Create a CSP model with three variables:
- Variables: A, B, C
- Domain: {0,1,2,3}

Apply the following constraints:

1. A ≠ B
2. B ≠ C
3. A + B ≤ 4

Requirements:
- Define variables using model.new_int_var()
- Add constraints using model.add()
- Solve the model
- Print the values of A, B, and C

````python
from ortools.sat.python import cp_model

model = cp_model.CpModel()

a = model.new_int_var(0, 3, 'A')
b = model.new_int_var(0, 3, 'B')
c = model.new_int_var(0, 3, 'C')

model.add(a!=b)
model.add(b!=c)
model.add(a+b<=4)

solver = cp_model.CpSolver()

status = solver.Solve(model)

if status==cp_model.OPTIMAL or status==cp_model.FEASIBLE:
    print(f'A:{solver.value(a)}, B:{solver.value(b)}, C:{solver.value(c)}')
else:
    print("Solution not found")
````

# **Task 6 Simple Optimization Problem**
Create a CSP model with:
- Variables: x, y, z
- Domain: 0–20
- Constraints:
- - x + 2y + z ≤ 20
- - 3x + y ≤ 18
- Objective Function:
- - maximize 4x + 2y + z
- Requirements:
- - Define variables
- - Add constraints
- - Define the objective function
- -Solve the model
- Print:
 - -Optimal value
- - Values of x, y, and z

````python
from ortools.sat.python import cp_model

model = cp_model.CpModel()

x = model.new_int_var(0, 20, 'x')
y = model.new_int_var(0, 20, 'y')
z = model.new_int_var(0, 20, 'z')

model.add(x+2*y+z<=20)
model.add(3*x+y<=18)

model.maximize(4*x+2*y+z)

solver=cp_model.CpSolver()

status=solver.solve(model)

if status==cp_model.OPTIMAL or status==cp_model.FEASIBLE:
    print("Optimal Solution")
    print(f"x:{solver.value(x)}, y:{solver.value(y)}, z:{solver.value(z)}")
else:
    print("No solution found")
````

# **Task 7 N-Queens Variation**
Write a Python program using OR-Tools to solve the 4-Queens problem.

Requirements:
1. No two queens share the same row
2. No two queens share the same diagonal
3. Print at least one valid solution
Example output format:
```
_ Q _ _
_ _ _ Q
Q _ _ _
_ _ Q _
```



````python
from ortools.sat.python import cp_model

class printQueen(cp_model.CpSolverSolutionCallback):
    def __init__(self, variables:list[cp_model.IntVar]):
        cp_model.CpSolverSolutionCallback.__init__(self)
        self.__variables=variables
        self.__count=0
    def on_solution_callback(self)->None:
        self.__count+=1
        print(f"Solution {self.__count}:")
        arr=[]
        for v in self.__variables:
            arr.append(self.value(v))

        for i in range(4):
            for j in range(4):
                if (arr[i]==j):
                    print('Q', end=' ')
                else:
                    print('-', end=' ')
            print()
        print()

model=cp_model.CpModel()
solver=cp_model.CpSolver()

a=model.new_int_var(0,3,'a')
b=model.new_int_var(0,3,'b')
c=model.new_int_var(0,3, 'c')
d=model.new_int_var(0, 3, 'd')

model.add(a!=b)
model.add(a!=c)
model.add(a!=d)
model.add(b!=c)
model.add(b!=d)
model.add(c!=d)

diff1=model.new_int_var(0, 3, 'diff1')
diff2=model.new_int_var(0, 3, 'diff2')
diff3=model.new_int_var(0, 3, 'diff3')
diff4=model.new_int_var(0, 3, 'diff4')
diff5=model.new_int_var(0, 3, 'diff5')
diff6=model.new_int_var(0, 3, 'diff6')

model.add_abs_equality(diff1, a-b)
model.add_abs_equality(diff2, a-c)
model.add_abs_equality(diff3, a-d)
model.add_abs_equality(diff4, b-c)
model.add_abs_equality(diff5, b-d)
model.add_abs_equality(diff6, c-d)

model.add(diff1!=1)
model.add(diff2!=2)
model.add(diff3!=3)
model.add(diff4!=1)
model.add(diff5!=2)
model.add(diff6!=1)

printer=printQueen([a,b,c,d])
solver.parameters.enumerate_all_solutions=True
solver.solve(model,printer)
````
