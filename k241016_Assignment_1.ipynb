### **Question 1:** Student Grade System

Write a Python program that takes a student’s name and marks as input and prints the grade
according to the following criteria:

• Marks ≥ 85 → Grade A

• Marks ≥ 70 → Grade B

• Marks ≥ 50 → Grade C

• Marks < 50 → Fail

```python
name = input("Enter student name: ")
marks = int(input("Enter marks: "))
grade = 'A'
if (marks>=85):
  grade = 'A'
elif (marks>=70):
  grade = 'B'
elif (marks>=50):
  grade = 'C'
else:
  grade = 'Fail'
print("Grade:", grade)
```

### **Question 2:** Even Number Counter

Write a Python program that takes a number n from the user and prints all even numbers from 1
to n. Also display the total count of even numbers.

```python
n = int(input("Enter a number: "))
print("Even numbers: ", end='')
count=0
for i in range(2, n+1,2):
  print(i, end=' ')
  count+=1
print("\nTotal even numbers:", count)
```

### **Question 3:** Student Record Using Dictionary

Write a Python program that stores the names and marks of three students in a dictionary and
then displays all student records.

```python
dict = {}
for i in range(3):
  name = input("Enter name: ")
  marks = int(input("Enter marks: "))
  dict[name] = marks

print("\nStudent Records:")
for name in dict.keys():
  marks = dict[name]
  print(name, ":", marks, end="\n")
```

### **Question 4:** Menu-Driven Calculator

Write a Python program using a while loop that shows the following menu repeatedly until the
user exits:
1. Add two numbers
2. Subtract two numbers
3. Exit

```python
while (1):
  print("1. Add two numbers")
  print("2. Subtract two numbers")
  print("3. Exit")
  choice = int(input("Enter your choice: "))
  if (choice == 3):
    print("Exiting")
    break
  num1 = int(input("Enter first number: "))
  num2 = int(input("Enter second number: "))
  if (choice == 1):
    num1+=num2
  else:
    num1-=num2
  print("Result:", num1, "\n")
```

### **Question 5:** Average Marks Using Function
Question:
Write a Python function named calculate_average() that takes a list of marks and returns the
average marks.

```python
def calculate_average(marks):
  total=0
  for mark in marks:
    total+=mark
  return float(total/len(marks))

#As question only asked to create the function, i have hardcoded the list
#i have provided without hardcored version below
marks=[70, 80, 90]
print("Marks: [70, 80, 90]")
print("Average Marks:", calculate_average(marks))
```
