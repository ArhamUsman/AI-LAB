#**Task 1** – Hill Climbing<br>
Use the Hill Climbing algorithm to find the maximum value of the function<br>
`f(x) = -x2 + 6x`<br>

**Range:** 0 ≤ x ≤ 6
<br>

**Instructions:**
- Start from a random value of x.
- Check the neighbor values (x+1 and x−1).
- Move to the neighbor with the higher value of f(x).
- Stop when no better neighbor exists.

**Output should show:**
- Initial value of x
- Value of f(x) at each step
- Final optimal value.

```python
import random

def calculate_val(x):
  return -(x*x)+(6*x)

def get_neighbours(x):
  neighbours=[]
  if (x-1>=0):
    neighbours.append(x-1)
  if (x+1<=6):
    neighbours.append(x+1)
  return neighbours

def hill_climbing():
  x=random.randint(0,6)
  print(f"Initial value of x = {x}")
  while True:
    neighbours=get_neighbours(x)
    next_state=x
    for neighbour in neighbours:
      if (calculate_val(neighbour)>calculate_val(next_state)):
        next_state=neighbour
    if next_state==x:
      break
    x=next_state
    print(f"value of x: {x}, f(x): {calculate_val(x)}")
  return x, calculate_val(x)

x=hill_climbing()
print(f"Final optimal value {x[1]} at x={x[0]}")
```

#**Task 2** – Beam Search
Use Beam Search to reach the number 20 starting from 1.
<br>
Allowed operations:
- `x → x + 2`
- `x → x + 3`
- `x → x × 2`

**Heuristic function:**

`h(n) = |20 − current number|`
<br>

**Beam width:** `k = 2`

**Instructions:**
- Start from `number 1`.
- Generate the next possible numbers using the allowed operations.
- At each level keep only the `best 2` states with the lowest heuristic value.
- Continue until the `goal number 20` is reached.

**Output should show:**
- Numbers explored at each level
- Final path to reach 20.

```python
import heapq
from typing import List, Tuple
def heuristic(x):
  h=20-x
  if h<0:
    h*=-1
  return h
def beam_search(start, goal, width=2):
  beam = [(heuristic(start), [start])]
  while beam:
    candidates=[]
    for h, path in beam:
      x=path[-1]
      print(f"number explored: {x}")
      if x==goal:
        print("Goal found")
        return path
      else:
        for neighbour in [x+2, x+3, x*2]:
          new_path = path + [neighbour]
          candidates.append((heuristic(neighbour), new_path))
    beam=heapq.nsmallest(width, candidates, key=lambda x:x[0])
    print(f"beam: {beam}")
    print("-----------------------------\n")

x=beam_search(1, 20)
print(f"Final path: {x}")
```

#**Task 3** – Genetic Algorithm

Use a *Genetic Algorithm* to maximize the function `f(x) = x2 + 2x`

**Range:** `0 ≤ x ≤ 31`

**Instructions:**
- Use binary representation for chromosomes.
- Generate a random initial population.
- Calculate fitness using the function `f(x)`.
- Apply selection, crossover, and mutation.
- Run the algorithm for `15` generations.

**Output should show:**
- Best chromosome
- Best value of x
- Best fitness value.

```python
import random

def fitness(chromosome):
  x = int(chromosome, 2)
  return x*x+2*x

def selection(pop, fitness_vals):
  tot_fitness=sum(fitness_vals)
  prob=[fitness/tot_fitness for fitness in fitness_vals]
  selected=random.choices(pop, prob, k=2)
  print(f"selected: {selected}")
  return selected

def crossover(p1, p2):
  crossover_point=random.randint(1, len(p1)-1)
  child1=p1[:crossover_point]+p2[crossover_point:]
  child2=p2[:crossover_point]+p1[crossover_point:]
  return child1, child2

def mutate(chromosome, mutation_rate):
  chromosome = list(chromosome)
  for i in range(len(chromosome)):
    if random.random()<mutation_rate:
      if chromosome[i]=='0':
        chromosome[i]='1'
      else:
        chromosome[i]='0'
  return "".join(chromosome)

def genetic_algorithm(initial_pop, mutation_rate, num_generations):
  print(f"Initial pop: {initial_pop}")
  pop=initial_pop
  for generation in range(num_generations):
    fitness_vals=[fitness(chromosome) for chromosome in pop]
    parents=selection(pop, fitness_vals)
    offsprings=[]
    for i in range(len(pop)//2):
      c1, c2=crossover(parents[0], parents[1])
      offsprings.append(c1)
      offsprings.append(c2)
    #offsprings=[gene for sublist in offsprings for gene in sublist]
    print(f"offsprings: {offsprings}")
    offsprings=[mutate(chromosome, mutation_rate) for chromosome in offsprings]
    print(f"mutated offsprings: {offsprings}")
    pop=offsprings
    print("\n----------------------------\n")
    print(f"Generation {generation+1}: {pop}")

  best_chromosome=max(pop, key=fitness)
  return best_chromosome

initial_pop=[]
for i in range(6):
  x=''
  for j in range(5):
    x+=random.choice('01')
  initial_pop.append(x)

best_chromosome=genetic_algorithm(initial_pop, 0.1, 15)
print(f"Best chromosome: {best_chromosome}")
print(f"Best value of x: {int(best_chromosome, 2)}")
```
